# 2. Treine e teste o modelo para cada região em geo_data_0.csv:


## 2.1 Divida os dados em um conjunto de treinamento e um conjunto de validação em uma proporção de 75:25.


```python
def split_region(df):
    features = df[['f0', 'f1', 'f2']]
    target = df['product']

    return train_test_split(features, target, test_size=0.25, random_state=12345)

features_train_0, features_valid_0, target_train_0, target_valid_0 = split_region(geo_data_0)
features_train_1, features_valid_1, target_train_1, target_valid_1 = split_region(geo_data_1)
features_train_2, features_valid_2, target_train_2, target_valid_2 = split_region(geo_data_2)
```

<br><br>
Optou-se por fazer a divisão dos dados de maneira individual para cada DataFrame, pois misturar os dados invalidaria a avaliação e poderia levar a decisões erradas. Com a divisão individual evita-se viés e reflete corretamente o comportamento real dos dados. Também por se tratar de regiões diferentes.

## 2.2 Treine o modelo e faça predições para o conjunto de validação.


## 2.3 Salve as predições e respostas corretas no conjunto de validação.


## 2.4 Imprima o volume médio previsto de reservas e o REQM do modelo.


```python
# Treinamento de modelo LinearRegression.

model_0 = LinearRegression().fit(features_train_0, target_train_0)
model_1 = LinearRegression().fit(features_train_1, target_train_1)
model_2 = LinearRegression().fit(features_train_2, target_train_2)

# Predição para conjunto de validação.

predicted_valid_0 = model_0.predict(features_valid_0)
predicted_valid_1 = model_1.predict(features_valid_1)
predicted_valid_2 = model_2.predict(features_valid_2)

# Calculo MEDIA

mse_0 = mean_squared_error(target_valid_0, predicted_valid_0)
mse_1 = mean_squared_error(target_valid_1, predicted_valid_1)
mse_2 = mean_squared_error(target_valid_2, predicted_valid_2)
print('Região 0: \nEQM: ', mse_0, '\nREQM: ', (mse_0 ** 0.5))
print(),
print('Região 1: \nEQM: ', mse_1, '\nREQM: ', (mse_1 ** 0.5))
print()
print('Região 2: \nEQM: ', mse_2, '\nREQM: ', (mse_2 ** 0.5))
```

# 4. Escreva uma função para calcular lucro de um conjunto de poços de petróleo selecionados e predições do modelo:


## 4.1 Escolha os 200 poços com os valores mais altos previstos de cada uma das 3 regiões (ou seja, arquivos 'csv').
## 4.2 Sumarize o volume alvo de reservas de acordo com essas predições. Armazene as predições para os 200 poços para cada uma das 3 regiões.
## 4.3 Calcule o lucro potencial dos 200 melhores poços por região. Apresente suas conclusões: sugira uma região para o desenvolvimento de poços de petróleo e justifique sua escolha.

```python
INVESTIMENTO_TOTAL = 100_000_000  # dólares
N_POCOS = 200
RECEITA_POR_UNIDADE = 4500  # dólares por unidade de reserva

def calcular_lucro(y_true, y_pred):
   
    # Criar DataFrame com valores reais e previstos
    dados = pd.DataFrame({
        'real': y_true,
        'predito': y_pred
    })
    
    # Selecionar os 200 poços com maiores valores previstos
    melhores_pocos = dados.sort_values(
        by='predito',
        ascending=False
    ).head(N_POCOS)
    
    # Soma real das reservas dos 200 poços selecionados
    volume_total = melhores_pocos['real'].sum()
    
    # Receita total
    receita_total = volume_total * RECEITA_POR_UNIDADE
    
    # Lucro
    lucro = receita_total - INVESTIMENTO_TOTAL
    
    return lucro, volume_total, melhores_pocos
```

```python
# Região 0
lucro_0, volume_0, top_0 = calcular_lucro(
    target_valid_0,
    predicted_valid_0
)

# Região 1
lucro_1, volume_1, top_1 = calcular_lucro(
    target_valid_1,
    predicted_valid_1
)

# Região 2
lucro_2, volume_2, top_2 = calcular_lucro(
    target_valid_2,
    predicted_valid_2
)
```

# 5. Calcule riscos e lucro para cada região:


## 5.1 Usando as predições que você armazenou na etapa 4.2, use a técnica de bootstrapping com 1.000 amostras para encontrar a distribuição de lucros.
## 5.2 Encontre lucro médio, intervalo de confiança de 95% e o risco de prejuízo. Prejuízo é um lucro negativo, calcule-o como uma probabilidade e depois o expresse como uma porcentagem.
## 5.3Apresente suas conclusões: sugira uma região para o desenvolvimento de poços de petróleo e justifique sua escolha. A sua escolha corresponde à escolha anterior na etapa 4.3?

```python
# Atribuição de variáveis

INVESTIMENTO_TOTAL = 100_000_000
RECEITA_POR_UNIDADE = 4500
N_POCOS = 200
N_BOOTSTRAP = 1000
```

```python
# Função cálculo de lucro
def calcular_lucro(y_true, y_pred):
    dados = pd.DataFrame({
        'real': y_true.reset_index(drop=True),
        'predito': y_pred
    })

    top_200 = dados.sort_values(
        by='predito',
        ascending=False
    ).head(N_POCOS)

    volume_total = top_200['real'].sum()
    receita_total = volume_total * RECEITA_POR_UNIDADE
    lucro = receita_total - INVESTIMENTO_TOTAL

    return lucro
```

```python
## Bootstrapping para distribuição de lucros
def bootstrap_lucro(y_valid, y_pred, n_bootstrap=1000):
    lucros = []

    dados = pd.DataFrame({
        'real': y_valid.reset_index(drop=True),
        'predito': y_pred
    })

    for _ in range(n_bootstrap):
        amostra = dados.sample(
            n=500,
            replace=True
           
        )

        amostra_ordenada = amostra.sort_values(
            by='predito',
            ascending=False
        ).head(N_POCOS)

        volume = amostra_ordenada['real'].sum()
        receita = volume * RECEITA_POR_UNIDADE
        lucro = receita - INVESTIMENTO_TOTAL

        lucros.append(lucro)

    return pd.Series(lucros)
```

```python
#Aplicando o bootstrap por região

lucros_0 = bootstrap_lucro(target_valid_0, predicted_valid_0)
lucros_1 = bootstrap_lucro(target_valid_1, predicted_valid_1)
lucros_2 = bootstrap_lucro(target_valid_2, predicted_valid_2)
```

```python

# Estatísticas: lucro médio, IC 95% e risco de prejuízo

def resumo_lucro(lucros):
    lucro_medio = lucros.mean()
    ic_inferior = lucros.quantile(0.025)
    ic_superior = lucros.quantile(0.975)
    risco_prejuizo = (lucros < 0).mean() * 100

    return lucro_medio, ic_inferior, ic_superior, risco_prejuizo


res_0 = resumo_lucro(lucros_0)
res_1 = resumo_lucro(lucros_1)
res_2 = resumo_lucro(lucros_2)

```
