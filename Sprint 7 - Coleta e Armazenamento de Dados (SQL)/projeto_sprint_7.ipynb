# Passo 5. Testando hipóteses (Python)

```python
trips = pd.read_csv('project_sql_result_07.csv')

trips.info()
print()
print(trips)
```

## Teste de Hipótese

#### H0 - A duração média dos passeios do Loop para o Aeroporto Internacional O'Hare são iguais nos sábados chuvosos.
#### H1 - A duração média dos passeios do Loop para o Aeroporto Internacional O'Hare muda nos sábados chuvosos.

```python
import pandas as pd
from scipy import stats

import pandas as pd
from scipy import stats

trips['start_ts'] = pd.to_datetime(trips['start_ts'])

# Criar coluna com dia da semana
trips['weekday'] = trips['start_ts'].dt.day_name()

# Filtrar apenas sábados
trips_sat = trips[trips['weekday'] == 'Saturday']

good = trips_sat[trips_sat['weather_conditions'] == 'Good']['duration_seconds']
bad = trips_sat[trips_sat['weather_conditions'] == 'Bad']['duration_seconds']


equal_var = good.var() == bad.var()

t_stat, p_value = stats.ttest_ind(good, bad, equal_var=equal_var)

print("Média Good:", good.mean())
print("Média Bad:", bad.mean())
print("p-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("\nConclusão: Rejeitamos H0. A duração média muda em sábados chuvosos.")
else:
    print("\nConclusão: Não rejeitamos H0. A duração média não muda em sábados chuvosos.")
```

```python
print(bad)
print()
print('Existem {} viagens em tempo Bad.'.format(bad.count()))
```

Apesar de haver significativamente menos viagens registradas sob a condição climática Bad, a média de duração dessas corridas mostrou-se substancialmente superior quando comparada ao restante das viagens.

***Análise de método***

Para essa análise, empregou-se o critério estatístico padrão, o Teste t de Student para duas amostras independentes, adotando-se, por convenção, um nível de significância (α) igual a 0,05.

A escolha desse teste é justificada pelo fato de a variável analisada ser contínua, pela existência de dois grupos distintos para comparação e pela independência entre as observações — uma vez que cada viagem representa um registro único e não relacionado às demais.
